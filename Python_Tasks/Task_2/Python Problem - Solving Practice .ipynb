#Q1 Find the shortest Word in a sentence

->3)take a list and assume the first word as the shortest-> 4)compare the length of the remaining words with shorset word using a for loop5)->if a word is shorter than current shortest word update it

```python
sentence="Python makes coding easy"
words=sentence.split()
print(words)
shortest=words[0]
for i in words:
    if len(i)<len(shortest):
        shortest=i
print(shortest)
```

Q9. Find the Least Frequent Character
1)Take a string as input->)Take another empty string and count as 0 ->3)pick one character and compare its count with remaining characters->4)if count is shorter update  it as least frequent character and count.

```python
string="success"
min_char=" "
min_count=len(string)
for ch in string:
    count=0
    for i in string:
        if ch==i:
            count+=1
    if count<min_count:
        min_count=count
        min_char=ch
        
print("Least frequent character:", min_char)
```

```python
string = "programming"
result = []

for ch in string:
    count = 0

    for i in string:
        if ch == i:
            count += 1

    if count > 1 and ch not in result:
        result.append(ch)

print(result)
```

Q17. Find All Words Appearing More Than Once
1)Take a string and split it->2)take an empty list as repeated ->3)using for loop count the words in list and using if condition compare counts  ->4)if count is more than 1 append words to repaeted and print list

```python
sentence ="data science uses data and science tools"
words = sentence.split()
repeated = []

for word in words:
    if words.count(word) > 1 and word not in repeated:
        repeated.append(word)

print(repeated)
```

Q19. Count Words Whose Length Is Greater Than 4
1)Take a string and split it->2)take count = 0->3)using for loop count the words in list and using if condition check for words whose length is greater than 4  -> 4)if length is more than 4 print the count of words whose length is greater than 4

```python
sentence = "Python makes problem solving interesting"
words = sentence.split()
count = 0

for word in words:
    if len(word) > 4:
        count += 1

print(count)
```

Q20. Find the Longest Word That Appears Only Once
1)Take a string and split it-> 2)take an empty string as longest->3)using for loop compare length of words and using if condition count the appearance of words ->4)if the word's count is 1 & longest update and print it


```python
sentence = "Python makes coding useful Python".lower()
words = sentence.split()
longest =""

for word in words:
    if words.count(word) == 1 and len(word) > len(longest):
        longest = word

print(longest)
```
