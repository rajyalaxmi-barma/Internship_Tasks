## Q1. Frequency of Each Word in a Sentence
**Pseudocode**

1. Create an empty dictionary.
2. Split the sentence into words.
3. For each word, add 1 to its count (start at 0 if new).
4. Return the dictionary.

**Example input:** `python is easy and python is powerful`  
**Expected output:** `{'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}`

```python
def word_frequency(sentence):
    freq = {}
    for word in sentence.split():
        freq[word] = freq.get(word, 0) + 1
    return freq

print(word_frequency("python is easy and python is powerful"))
print(word_frequency("to be or not to be"))   # extra test
```

## Q2. Frequency of Each Character in a Word
**Pseudocode**

1. Create an empty dictionary.
2. Go through each character of the word.
3. Increase that character's count by 1.
4. Return the dictionary.

**Example input:** `banana`  
**Expected output:** `{'b': 1, 'a': 3, 'n': 2}`

```python
def char_frequency(word):
    freq = {}
    for ch in word:
        freq[ch] = freq.get(ch, 0) + 1
    return freq

print(char_frequency("banana"))
print(char_frequency("hello"))   # extra test
```

## Q3. First Non-Repeating Character
**Pseudocode**

1. Count every character's frequency.
2. Walk through the word from the left.
3. The first character with count 1 is the answer.
4. If none found, return None.

**Example input:** `swiss`  
**Expected output:** `w`

```python
def first_non_repeating(word):
    freq = char_frequency(word)
    for ch in word:
        if freq[ch] == 1:
            return ch
    return None

print(first_non_repeating("swiss"))
print(first_non_repeating("aabb"))   # extra test -> None
```

## Q4. Check Whether Two Strings Are Anagrams
**Pseudocode**

1. Count character frequencies of both strings.
2. If both frequency dictionaries are equal, they are anagrams.
3. Otherwise they are not.

**Example input:** `s1 = "listen", s2 = "silent"`  
**Expected output:** `True`

```python
def are_anagrams(s1, s2):
    return char_frequency(s1) == char_frequency(s2)

print(are_anagrams("listen", "silent"))
print(are_anagrams("hello", "world"))   # extra test -> False
```

## Q5. First Repeating Character
**Pseudocode**

1. Keep an empty set of seen characters.
2. Read the string left to right.
3. If the character is already in the set, return it.
4. Otherwise add it to the set.
5. If the loop ends, return None.

**Example input:** `programming`  
**Expected output:** `r`

```python
def first_repeating(text):
    seen = set()
    for ch in text:
        if ch in seen:
            return ch
        seen.add(ch)
    return None

print(first_repeating("programming"))
print(first_repeating("abc"))   # extra test -> None
```

## Q6. Move All Zeros to the End
**Pseudocode**

1. Collect all non-zero elements in their original order.
2. Count how many zeros were removed.
3. Add that many zeros at the end.

**Example input:** `[0, 3, 0, 1, 2, 0, 4]`  
**Expected output:** `[3, 1, 2, 4, 0, 0, 0]`

```python
def move_zeros(nums):
    non_zeros = [n for n in nums if n != 0]
    return non_zeros + [0] * (len(nums) - len(non_zeros))

print(move_zeros([0, 3, 0, 1, 2, 0, 4]))
print(move_zeros([0, 0, 5]))   # extra test
```

## Q7. Second Highest Distinct Number
**Pseudocode**

1. Keep two variables: highest and second, both empty at first.
2. For each number: if it beats highest, old highest becomes second and number becomes highest.
3. Else if it differs from highest and beats second, it becomes second.
4. Return second (None if never set).

**Example input:** `[10, 20, 40, 30, 40, 20]`  
**Expected output:** `30`

```python
def second_highest(nums):
    highest = second = None
    for n in nums:
        if highest is None or n > highest:
            second = highest
            highest = n
        elif n != highest and (second is None or n > second):
            second = n
    return second

print(second_highest([10, 20, 40, 30, 40, 20]))
print(second_highest([5, 5, 5]))   # extra test -> None
```

## Q8. Remove Duplicates Preserving Order
**Pseudocode**

1. Keep a set of seen values and a result list.
2. For each element, if not seen, add to both set and result.
3. Return the result.

**Example input:** `[4, 2, 4, 5, 2, 7, 5]`  
**Expected output:** `[4, 2, 5, 7]`

```python
def remove_duplicates(nums):
    seen = set()
    result = []
    for n in nums:
        if n not in seen:
            seen.add(n)
            result.append(n)
    return result

print(remove_duplicates([4, 2, 4, 5, 2, 7, 5]))
print(remove_duplicates([1, 1, 1]))   # extra test
```

## Q9. Missing Number from 1 to N
**Pseudocode**

1. Expected total of 1..N is N*(N+1)/2.
2. Add up the given list.
3. The difference is the missing number.

**Example input:** `nums = [1, 2, 4, 5, 6], N = 6`  
**Expected output:** `3`

```python
def missing_number(nums, n):
    return n * (n + 1) // 2 - sum(nums)

print(missing_number([1, 2, 4, 5, 6], 6))
print(missing_number([2, 3, 4, 5], 5))   # extra test -> 1
```

## Q10. Two Numbers Whose Sum Equals a Target
**Pseudocode**

1. Scan left to right, remembering the earliest index of each value seen.
2. For the current number, compute need = target - current.
3. If need was seen earlier, return [earlier value, current].
4. Otherwise store current (if new) and continue.
5. If nothing found, return [].

**Example input:** `nums = [2, 7, 11, 15], target = 9`  
**Expected output:** `[2, 7]`

```python
def two_sum_pair(nums, target):
    first_seen = {}
    for j, current in enumerate(nums):
        need = target - current
        if need in first_seen:
            return [nums[first_seen[need]], current]
        if current not in first_seen:
            first_seen[current] = j
    return []

print(two_sum_pair([2, 7, 11, 15], 9))
print(two_sum_pair([1, 2, 3], 10))   # extra test -> []
```

## Q11. Number Closest to Zero
**Pseudocode**

1. Assume the first number is the best.
2. For each other number, replace best if its absolute value is smaller.
3. On a tie in absolute value, prefer the larger (positive) number.
4. Return best.

**Example input:** `[-8, -3, 2, 5, -2]`  
**Expected output:** `2`

```python
def closest_to_zero(nums):
    best = nums[0]
    for n in nums[1:]:
        if abs(n) < abs(best) or (abs(n) == abs(best) and n > best):
            best = n
    return best

print(closest_to_zero([-8, -3, 2, 5, -2]))
print(closest_to_zero([-4, 4, 9]))   # extra test -> 4
```

## Q12. Common Elements Between Two Lists
**Pseudocode**

1. Put list2 into a set for fast lookup.
2. Go through list1 in order.
3. If an item is in list2 and not already added, add it to the result.
4. Return result.

**Example input:** `list1 = [1, 2, 3, 4, 5], list2 = [3, 4, 5, 6, 7]`  
**Expected output:** `[3, 4, 5]`

```python
def common_elements(list1, list2):
    in_second = set(list2)
    added = set()
    result = []
    for x in list1:
        if x in in_second and x not in added:
            added.add(x)
            result.append(x)
    return result

print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))
print(common_elements([1, 1, 2, 3], [3, 1, 1]))   # extra test -> [1, 3]
```

## Q13. Rotate a List Right by K
**Pseudocode**

1. If the list is empty, return [].
2. Reduce K with K % length so large K still works.
3. Take the last K items and put them in front of the remaining items.

**Example input:** `nums = [1, 2, 3, 4, 5], K = 2`  
**Expected output:** `[4, 5, 1, 2, 3]`

```python
def rotate_right(nums, k):
    if not nums:
        return []
    k %= len(nums)
    if k == 0:
        return nums[:]
    return nums[-k:] + nums[:-k]

print(rotate_right([1, 2, 3, 4, 5], 2))
print(rotate_right([1, 2, 3], 4))   # extra test -> [3, 1, 2]
```

## Q14. Maximum Sum of Two Consecutive Elements
**Pseudocode**

1. Start best as the sum of the first two elements.
2. Slide through the list, adding each element to its next neighbour.
3. Keep the larger sum.
4. Return best.

**Example input:** `[4, 7, 2, 9, 5]`  
**Expected output:** `14`

```python
def max_consecutive_sum(nums):
    best = nums[0] + nums[1]
    for i in range(1, len(nums) - 1):
        best = max(best, nums[i] + nums[i + 1])
    return best

print(max_consecutive_sum([4, 7, 2, 9, 5]))
print(max_consecutive_sum([-5, -1, -3]))   # extra test -> -4
```

## Q15. All Pairs Whose Sum Equals a Target
**Pseudocode**

1. Count how often each number occurs.
2. Go through the distinct numbers in ascending order.
3. For number a, partner b = target - a.
4. If a < b and b exists, add [a, b].
5. If a == b, add [a, a] only if a occurs at least twice.
6. Sorted iteration keeps pairs ordered by first element.

**Example input:** `nums = [2, 4, 3, 5, 7, 8, 1], target = 9`  
**Expected output:** `[[1, 8], [2, 7], [4, 5]]`

```python
def all_pairs(nums, target):
    counts = char_frequency(nums)
    pairs = []
    for a in sorted(counts):
        b = target - a
        if a < b and b in counts:
            pairs.append([a, b])
        elif a == b and counts[a] >= 2:
            pairs.append([a, b])
    return pairs

print(all_pairs([2, 4, 3, 5, 7, 8, 1], 9))
print(all_pairs([3, 3, 3, 6], 6))   # extra test -> [[3, 3]]
```
